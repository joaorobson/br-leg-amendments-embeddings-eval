```python
default_header_pattern =  r"^[\s\S]*?((PEC\s+n\.º\s+6\s+de\s+2019|À\s+Proposta\s+de\s+Emenda\s+à\s+Constituição\s+n[º°o]\s*6\/2019\s+que)\s+modifica\s+o\s+sistema\s+de\s+previdência\s+social[\s\S]+?provid[eê]ncias)|(\(à\s+PEC\s+n[º°o]\s*6,\s*2019\)|\(à\s+PEC\s+n[º°o]\s*6,\s*de\s*2019\)|\(à\s+Proposta\s+de\s+Emenda\s+à\s+Constituição\s+n[º°o]\s*6,\s*de\s*2019\)\s+(Aditiva|Modificativa)?|À\s+PEC\s+N[º°o]\s*06,\s*de\s*2019\.\s*\n\s*\([^\n]+\)|EMENDA\s+N[º°o]\s+CCJ\s+\(PEC\s6\/2019\s+|EMENDA\s+SUPRESSIVA(?:\s*-\s*CCJ|\s+À\s+PEC\s+06/2019\.)?|EMENDA\s+MODIFICATIVA(\s+Nº)?|EMENDA\s+ADITIVA|EMENDA\s+N[º°o\.]*\s*-\s*CCJ\s*\n?\s*(\(à\s+PEC\s+n[º°o]\s*6,\s*de\s*2019\))?(\s*EMENDA.*?\s+À\s+PEC\s+06/2019\.)?|EMENDA\s+N\.º\s*_______\/CCJ|\(PEC\s*6\/2019\)|\(Do\s+Sr\.\s+Randolfe\s+Rodrigues\s+e\s+outros\))"
default_header_regex = re.compile(default_header_pattern, re.VERBOSE | re.IGNORECASE)
```

```python
print(default_header_pattern)
```

```python
def split_header(text, regex=default_header_regex):
    if pd.isna(text):
        return pd.Series([None, None])

    match = regex.search(text)

    if match:
        fim = match.end()

        # inclui o termo encontrado no cabeçalho
        cabecalho = text[:fim].strip()

        # restante após o termo
        texto_sem_cabecalho = text[fim:].strip()

        return pd.Series([cabecalho, texto_sem_cabecalho])

    return pd.Series([None, text])
```

```python
pec[["cabecalho", "texto_sem_cabecalho"]] = (
    pec["texto"]
    .apply(split_header)
)
```

### Remover dados do corpo

#### Remover gabinete

```python
print(len(pec[pec.texto_sem_cabecalho.str.contains("Gabinete d")]))
```

```python
trechos_com_gabinete = (
    pec.loc[
        pec["texto_sem_cabecalho"].str.contains("Gabinete d", na=False),
        "texto_sem_cabecalho"
    ]
    .str.extract(r"(.{0,20}Gabinete d.{0,20})", expand=False)
)

print(trechos_com_gabinete.tolist())
```

```python
def remove_gabinete(text):
    pattern = r'([\s\r\n]*Senado Federal[\s\r\n]*)?(?:Gabinete\s*(?:d[oa]\s*)?Senadora?\s+[a-zA-Zà-úÀ-Ú ]+[\s\r\n]+)'
    return re.sub(pattern, '', text, flags=re.IGNORECASE | re.MULTILINE)
```

```python
mask = pec["texto_sem_cabecalho"].str.contains("Gabinete d", na=False)

# linhas com "Gabinete d"
pec.loc[mask, "texto_preprocessado"] = (
    pec.loc[mask, "texto_sem_cabecalho"]
       .apply(remove_gabinete)
)

# linhas sem "Gabinete d"
pec.loc[~mask, "texto_preprocessado"] = (
    pec.loc[~mask, "texto_sem_cabecalho"]
)

trechos_restantes = (
    pec.loc[
        pec["texto_preprocessado"].str.contains("Gabinete d", na=False),
        "texto_preprocessado"
    ]
    .str.findall(r".{0,20}Gabinete d.{0,20}")
)

print(trechos_restantes.tolist())
```

```python
def remove_address(text):
    pattern = r'(?i)(?:S\s*E\s*N\s*A\s*D\s*O\s+F\s*E\s*D\s*E\s*R\s*A\s*L|Senado\s+Federal|Praça\s+dos\s+Três\s+Poderes)(?:[\s,|–-]*(?:Anexo\s+[IVX0-9]+|Ala\s+(?:Senador\s+)?[A-Za-zÀ-ÿ]+(?:\s+[A-Za-zÀ-ÿ]+){0,3}|Gabinete\s+(?:do\s+)?Senador[a]?\s+[A-Za-zÀ-ÿ]+(?:\s+[A-Za-zÀ-ÿ]+){0,2}|Gabinete\s+\d+|Gab\.\s*\d+|\d+º\s+Andar|Térreo|CEP\s*\d{5}(?:[-.]\d{3})?\.?|\d{5}[-.]\d{3}|Brasília(?:\s*[-–|I]*\s*DF)?|Fones?\s*[\d\s.–-]+|Telefones?:\s*\+?\d*|\bBra\b|\bGabine\b|Praça\s+dos\s+Três\s+Poderes|Senado\s+Federal)){1,10}'
    return re.sub(pattern, '', text, flags=re.IGNORECASE | re.MULTILINE)
```

```python
pec["texto_preprocessado"] = pec.texto_preprocessado.apply(remove_address)
```

```python
trechos_com_endereco = (
    pec.loc[
        pec.texto_preprocessado.str.contains(r"Praça\s|CEP\s|Ala\s|Gabinete\s", na=False),
        "texto_preprocessado"
    ]
    # Regex atualizada que aceita dinamicamente os dois casos
    .str.extractall(r"([\s\S]{0,80}(Praça\s|CEP\s|Ala\s|Gabinete\s)[\s\S]{0,80})")
)

print(trechos_com_endereco[0].tolist())
```
