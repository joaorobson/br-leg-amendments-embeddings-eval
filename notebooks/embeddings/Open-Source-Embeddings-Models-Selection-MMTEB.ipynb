```python
import pandas as pd
import mteb
```

```python
benchmark_scores_df = pd.read_csv("../../data/datasets/mteb_22_08_2026.csv")
benchmark_scores_df.head(30)
```

```python
benchmark_scores_df = benchmark_scores_df[(~benchmark_scores_df.Clustering.isna()) & (~benchmark_scores_df.Retrieval.isna())].reset_index(drop=True)
```

```python
benchmark_scores_df
```

```python
import pandas as pd

cols = ["Clustering", "Retrieval"]

df_borda = benchmark_scores_df.copy()

mask = (
    df_borda["Clustering"].isna()
    | df_borda["Retrieval"].isna()
)

df_borda = (
    df_borda.loc[~mask]
    .reset_index(drop=True)
)

# =====================================================
# GEMINI
# =====================================================
# https://arxiv.org/pdf/2605.27295
nova_linha = {
    "Model": "Gemini 2",
    "Clustering": 55.3,
    "Retrieval": 70.0,
}

df_borda = pd.concat(
    [df_borda, pd.DataFrame([nova_linha])],
    ignore_index=True
)

# =====================================================
# BORDA
# =====================================================

n = len(df_borda)

for col in cols:

    rank = df_borda[col].rank(
        ascending=False,
        method="min"
    )

    df_borda[f"{col}_borda"] = n - rank + 1

df_borda["Borda"] = df_borda[
    [f"{c}_borda" for c in cols]
].sum(axis=1)

# =====================================================
# LOOKUP MTEB
# =====================================================

model_lookup = {}

for m in mteb.get_model_metas():

    model_lookup[m.name] = m.open_weights

    # alguns rankings usam apenas o nome curto
    #model_lookup[m.name.split("/")[-1]] = m.open_weights

# =====================================================
# ABERTO / FECHADO
# =====================================================

df_borda["Aberto"] = (
    df_borda["Model"]
    .map(model_lookup)
)

# Gemini não existe no catálogo MTEB
df_borda.loc[
    df_borda["Model"] == "Gemini 2",
    "Aberto"
] = False

# =====================================================
# EMPRESA / MODELO
# =====================================================

df_borda["Empresa"] = (
    df_borda["Model"]
    .astype(str)
    .str.split("/")
    .str[0]
)

df_borda["Modelo"] = (
    df_borda["Model"]
    .astype(str)
    .str.split("/")
    .str[-1]
)

# =====================================================
# LICENÇA
# =====================================================

df_borda["Licença"] = (
    df_borda["Aberto"]
    .map({
        True: "Aberto",
        False: "Fechado"
    })
)
```

```python
# =====================================================
# ORDENAÇÃO FINAL
# =====================================================

df_borda = df_borda.sort_values(
    "Borda",
    ascending=False
)

resultado = df_borda[
    [
        "Empresa",
        "Modelo",
        "Licença",
        "Total Params (B)",
        "Clustering",
        "Retrieval",
        "Borda",
    ]
]

resultado = resultado.reset_index(drop=True)
resultado.index = resultado.index + 1
```
