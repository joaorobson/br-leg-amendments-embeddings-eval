```python
import pandas as pd
```

```python
plp_68 = pd.read_parquet("../../data/datasets/PLP_68_2024_preprocessado.parquet")
mpv_612 = pd.read_parquet("../../data/datasets/MPV_612_2013_preprocessado.parquet")
pec_6 = pd.read_parquet("../../data/datasets/PEC_6_2019_preprocessado.parquet")
```

```python
plp_68_temas = pd.read_parquet("../../data/temas/PLP_68_2024.parquet")
mpv_612_temas = pd.read_parquet("../../data/temas/MPV_612_2013.parquet")
pec_6_temas = pd.read_parquet("../../data/temas/PEC_6_2019.parquet")
```

```python
plp_68_textos_e_temas = plp_68.merge(
    plp_68_temas[['num_emenda', 'tema', 'tema_macro', 'tema_nivel_2']],
    on='num_emenda',
    how='inner'
)
```

```python
pec_6_textos_e_temas = pec_6.merge(
    pec_6_temas[['num_emenda', 'tema']],
    on='num_emenda',
    how='inner'
)
```

```python
mpv_612_textos_e_temas = mpv_612.merge(
    mpv_612_temas[['num_emenda', 'tema']],
    on='num_emenda',
    how='inner'
)
```

```python
cols = ["num_emenda", "materia", "texto", "texto_preprocessado", "texto_preprocessado_sem_justificativa", "tema"]
cols_plp_68 = cols + ["tema_macro", "tema_nivel_2"]
```

```python
df_completo = pd.concat([plp_68_textos_e_temas[cols_plp_68], mpv_612_textos_e_temas[cols], pec_6_textos_e_temas[cols]])
```
